```python
# Step 0D: Validate existing analysis columns

print("Edit status values:")
display(df["is_edited"].value_counts(dropna=False))

print("\nMissing values:")
display(df[
    [
        "duration",
        "time_taken_by_user",
        "analysis_char_count",
        "user_task_cps"
    ]
].isna().sum().rename("missing_count"))

print("\nDuration and task-time validity:")
print("Non-positive durations:", (df["duration"] <= 0).sum())
print("Non-positive task times:", (df["time_taken_by_user"] <= 0).sum())

print("\nCheck existing character-count columns:")
display(
    df[
        [
            "user_text",
            "char_count",
            "char_count_excluding_blank_marker",
            "analysis_char_count",
            "user_task_cps"
        ]
    ].head(10)
)
```

```python
# Step 0E: Create clean prerequisite columns without overwriting originals

df["duration_clean"] = pd.to_numeric(df["duration"], errors="coerce")
df["task_time_clean"] = pd.to_numeric(
    df["time_taken_by_user"], errors="coerce"
)

df["edit_status_clean"] = (
    df["is_edited"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map({"yes": True, "no": False})
)

# Verify whether the existing character count matches our analysis definition
text = df["user_text"].fillna("").astype(str).str.strip()

expected_char_count = (
    text.where(text.ne("[blank]"), "")
    .str.replace(r"\s+", "", regex=True)
    .str.len()
)

print(
    "Rows where analysis_char_count differs:",
    (df["analysis_char_count"] != expected_char_count).sum()
)

print(
    "Rows where existing CPS differs from recalculated CPS:",
    (
        ~np.isclose(
            df["user_task_cps"],
            expected_char_count / df["task_time_clean"].where(
                df["task_time_clean"] > 0
            ),
            equal_nan=True
        )
    ).sum()
)

print("\nUnrecognized edit statuses:", df["edit_status_clean"].isna().sum())
print("Invalid durations:", (df["duration_clean"] <= 0).sum())
print(
    "Missing or non-positive task times:",
    (
        df["task_time_clean"].isna()
        | (df["task_time_clean"] <= 0)
    ).sum()
)
```

```python
# Step 1A: Calculate TER warning indicators

df["ter"] = (
    df["task_time_clean"] / df["duration_clean"]
)

df["flag_ter_below_1"] = (
    (df["ter"] < 1) &
    (df["duration_clean"] >= 2)
)

df["flag_ter_above_50"] = df["ter"] > 50

df["flag_task_time_above_1h"] = (
    df["task_time_clean"] > 3600
)

ter_flags = pd.Series({
    "TER < 1 and audio >= 2 seconds":
        df["flag_ter_below_1"].sum(),
    "TER > 50":
        df["flag_ter_above_50"].sum(),
    "Task time > 1 hour":
        df["flag_task_time_above_1h"].sum()
}, name="flagged_tasks")

display(ter_flags.to_frame())
```

```python
# Step 1B: Inspect task volume per user

user_task_counts = df.groupby("user_id").size()

print("Total unique users:", df["user_id"].nunique())
print("Median tasks per user:", user_task_counts.median())
print("Users with at least 20 tasks:", (user_task_counts >= 20).sum())
print("Users with at least 50 tasks:", (user_task_counts >= 50).sum())

display(
    user_task_counts.describe(
        percentiles=[0.25, 0.50, 0.75, 0.90, 0.95]
    ).to_frame(name="tasks_per_user")
)
```

```python
# Step 1C: Calculate per-user edit rates

user_edit_summary = (
    df.groupby("user_id")
    .agg(
        total_tasks=("user_id", "size"),
        edited_tasks=("edit_status_clean", "sum"),
        unedited_tasks=("edit_status_clean", lambda x: (~x).sum())
    )
    .reset_index()
)

user_edit_summary["edit_rate"] = (
    user_edit_summary["edited_tasks"]
    / user_edit_summary["total_tasks"]
)

# Only assess users with sufficient task history
user_edit_summary["eligible_20_tasks"] = (
    user_edit_summary["total_tasks"] >= 20
)

user_edit_summary["eligible_50_tasks"] = (
    user_edit_summary["total_tasks"] >= 50
)

print("Users with >=20 tasks:", user_edit_summary["eligible_20_tasks"].sum())
print("Users with >=50 tasks:", user_edit_summary["eligible_50_tasks"].sum())

print("\nUsers with >=20 tasks and edit rate <30%:")
print(
    (
        user_edit_summary["eligible_20_tasks"]
        & (user_edit_summary["edit_rate"] < 0.30)
    ).sum()
)

print("\nUsers with >=50 tasks and edit rate <15%:")
print(
    (
        user_edit_summary["eligible_50_tasks"]
        & (user_edit_summary["edit_rate"] < 0.15)
    ).sum()
)

display(
    user_edit_summary
    .sort_values("total_tasks", ascending=False)
    .head(10)
)
```

```python
# Step 1: Derive candidate thresholds from observed data

metrics = {
    "TER": df["ter"],
    "Task time (seconds)": df["task_time_clean"],
    "User-task CPS": df["user_task_cps"]
}

threshold_results = []

for name, series in metrics.items():
    x = series.replace([np.inf, -np.inf], np.nan).dropna()

    q01 = x.quantile(0.01)
    q05 = x.quantile(0.05)
    q25 = x.quantile(0.25)
    q50 = x.quantile(0.50)
    q75 = x.quantile(0.75)
    q95 = x.quantile(0.95)
    q99 = x.quantile(0.99)

    iqr = q75 - q25
    lower_fence = max(0, q25 - 1.5 * iqr)
    upper_fence = q75 + 1.5 * iqr

    threshold_results.append({
        "metric": name,
        "valid_tasks": len(x),
        "p01": q01,
        "p05": q05,
        "p25": q25,
        "median": q50,
        "p75": q75,
        "p95": q95,
        "p99": q99,
        "IQR_lower_fence": lower_fence,
        "IQR_upper_fence": upper_fence,
        "below_lower_fence": (x < lower_fence).sum(),
        "above_upper_fence": (x > upper_fence).sum()
    })

threshold_table = pd.DataFrame(threshold_results).set_index("metric")
display(threshold_table.round(3))

# Edit-rate distribution among users with sufficient task history
eligible_users = user_edit_summary.loc[
    user_edit_summary["total_tasks"] >= 20
].copy()

print("\nEligible users for edit-rate analysis:", len(eligible_users))

if len(eligible_users) > 0:
    display(
        eligible_users["edit_rate"]
        .describe(percentiles=[0.10, 0.25, 0.50, 0.75, 0.90])
        .to_frame("edit_rate")
        .round(3)
    )
```

```python
# Step 2: Compare CPS thresholds and inspect flagged examples

valid_cps = df.loc[
    (df["duration_clean"] > 5) &
    df["user_task_cps"].notna()
].copy()

candidate_thresholds = [
    valid_cps["user_task_cps"].quantile(0.95),
    valid_cps["user_task_cps"].quantile(0.99),
    valid_cps["user_task_cps"].quantile(0.995),
    valid_cps["user_task_cps"].quantile(0.999)
]

cps_threshold_table = pd.DataFrame({
    "threshold": candidate_thresholds
})

cps_threshold_table["flagged_tasks"] = [
    (valid_cps["user_task_cps"] > threshold).sum()
    for threshold in candidate_thresholds
]

cps_threshold_table["percentage"] = (
    100 * cps_threshold_table["flagged_tasks"] / len(valid_cps)
)

display(cps_threshold_table.round(3))

print("\nHighest CPS examples (audio duration > 5 seconds):")

display(
    valid_cps.nlargest(15, "user_task_cps")[
        [
            "user_id",
            "duration_clean",
            "task_time_clean",
            "analysis_char_count",
            "user_task_cps",
            "user_text"
        ]
    ]
)
```

```python
# Recalculate CPS thresholds on tasks with audio duration > 5 seconds

valid_cps = df.loc[
    (df["duration_clean"] > 5)
    & df["user_task_cps"].notna()
].copy()

percentiles = [0.95, 0.99, 0.995, 0.999]

rows = []

for p in percentiles:
    threshold = valid_cps["user_task_cps"].quantile(p)
    count = (valid_cps["user_task_cps"] > threshold).sum()

    rows.append({
        "percentile": f"{p:.1%}",
        "threshold_cps": threshold,
        "flagged_tasks": count,
        "percentage_of_eligible_tasks": 100 * count / len(valid_cps)
    })

display(pd.DataFrame(rows).round(3))

# Check whether high-CPS tasks have task time shorter than audio duration
high_cps = valid_cps[
    valid_cps["user_task_cps"] > valid_cps["user_task_cps"].quantile(0.99)
].copy()

high_cps["task_time_shorter_than_audio"] = (
    high_cps["task_time_clean"] < high_cps["duration_clean"]
)

print("Top 1% CPS tasks:", len(high_cps))
print(
    "Top 1% with task time shorter than audio:",
    high_cps["task_time_shorter_than_audio"].sum()
)

display(
    high_cps[
        [
            "duration_clean",
            "task_time_clean",
            "analysis_char_count",
            "user_task_cps",
            "task_time_shorter_than_audio"
        ]
    ].head(10)
)
```

```python
# Step 3: Clean CPS threshold comparison and overlap with TER warnings

valid_cps = df.loc[
    (df["duration_clean"] > 5) &
    df["user_task_cps"].notna()
].copy()

thresholds = {
    "P95": valid_cps["user_task_cps"].quantile(0.95),
    "P99": valid_cps["user_task_cps"].quantile(0.99),
    "P99.5": valid_cps["user_task_cps"].quantile(0.995),
    "P99.9": valid_cps["user_task_cps"].quantile(0.999)
}

rows = []

for label, threshold in thresholds.items():
    cps_flag = valid_cps["user_task_cps"] > threshold
    ter_flag = valid_cps["flag_ter_below_1"]

    rows.append({
        "percentile": label,
        "threshold_cps": threshold,
        "cps_flagged": int(cps_flag.sum()),
        "also_TER_below_1": int((cps_flag & ter_flag).sum()),
        "CPS_flags_with_TER_below_1_pct": (
            100 * (cps_flag & ter_flag).sum() / cps_flag.sum()
            if cps_flag.sum() else 0
        )
    })

display(pd.DataFrame(rows).round(3))
```

```python
# Step 4: Consolidate task-level warning signals

df["warning_timing"] = (
    df["flag_ter_below_1"] |
    df["flag_ter_above_50"] |
    df["flag_task_time_above_1h"]
)

df["warning_cps_support"] = (
    (df["duration_clean"] > 5) &
    (df["user_task_cps"] > valid_cps["user_task_cps"].quantile(0.99))
)

df["warning_count_primary"] = (
    df["flag_ter_below_1"].astype(int) +
    df["flag_ter_above_50"].astype(int) +
    df["flag_task_time_above_1h"].astype(int)
)

summary = pd.Series({
    "Tasks with any timing warning": df["warning_timing"].sum(),
    "Tasks with high-CPS supporting signal": df["warning_cps_support"].sum(),
    "Tasks with multiple timing flags": (
        df["warning_count_primary"] > 1
    ).sum()
}, name="task_count")

display(summary.to_frame())
```

```python
# Step 5: Inspect overlap between timing warning rules

timing_overlap = pd.crosstab(
    df["flag_ter_below_1"],
    df["flag_ter_above_50"],
    rownames=["TER < 1"],
    colnames=["TER > 50"]
)

display(timing_overlap)

print("\nOverlap with task time > 1 hour:")
display(
    pd.crosstab(
        df["flag_ter_above_50"],
        df["flag_task_time_above_1h"],
        rownames=["TER > 50"],
        colnames=["Task time > 1 hour"]
    )
)

print("\nSample timing-warning records:")
display(
    df.loc[
        df["warning_timing"],
        [
            "user_id",
            "duration_clean",
            "task_time_clean",
            "ter",
            "analysis_char_count",
            "user_task_cps",
            "flag_ter_below_1",
            "flag_ter_above_50",
            "flag_task_time_above_1h"
        ]
    ].head(10)
)
```

```python
# PART 1: Final warning-signal rules and summary

# 1. Derive thresholds from the observed dataset
valid_ter = df["ter"].replace([np.inf, -np.inf], np.nan).dropna()
valid_cps = df.loc[
    (df["duration_clean"] > 5) & df["user_task_cps"].notna(),
    "user_task_cps"
].replace([np.inf, -np.inf], np.nan).dropna()

ter_p95 = valid_ter.quantile(0.95)
ter_p99 = valid_ter.quantile(0.99)
cps_p99 = valid_cps.quantile(0.99)

# 2. Primary timing warning: task time shorter than audio
df["final_flag_shorter_than_audio"] = (
    (df["ter"] < 1) & (df["duration_clean"] >= 2)
)

# 3. Extreme TER warning: top 1% of observed TER
df["final_flag_extreme_ter"] = df["ter"] > ter_p99

# 4. CPS supporting diagnostic: top 1% for audio > 5 seconds
df["final_flag_extreme_cps"] = (
    (df["duration_clean"] > 5) &
    (df["user_task_cps"] > cps_p99)
)

# 5. Consolidated task-level review flag
df["final_flag_for_review"] = (
    df["final_flag_shorter_than_audio"] |
    df["final_flag_extreme_ter"]
)

# 6. Summarize thresholds and counts
final_summary = pd.DataFrame([
    {
        "signal": "TER < 1; audio >= 2 sec",
        "threshold": 1,
        "flagged_tasks": int(df["final_flag_shorter_than_audio"].sum())
    },
    {
        "signal": "TER above P99",
        "threshold": ter_p99,
        "flagged_tasks": int(df["final_flag_extreme_ter"].sum())
    },
    {
        "signal": "CPS above P99; audio > 5 sec",
        "threshold": cps_p99,
        "flagged_tasks": int(df["final_flag_extreme_cps"].sum())
    },
    {
        "signal": "Any primary warning",
        "threshold": np.nan,
        "flagged_tasks": int(df["final_flag_for_review"].sum())
    }
])

final_summary["percentage_of_all_tasks"] = (
    100 * final_summary["flagged_tasks"] / len(df)
)

print("TER P95:", round(ter_p95, 3))
print("TER P99:", round(ter_p99, 3))
print("CPS P99 (audio > 5 sec):", round(cps_p99, 3))

display(final_summary.round(3))

# 7. Edit-rate assessment: report evidence, don't force a threshold
eligible_users = user_edit_summary.loc[
    user_edit_summary["total_tasks"] >= 20
].copy()

print("\nEDIT-RATE ASSESSMENT")
print("Users with >=20 tasks:", len(eligible_users))
print(
    "Eligible users with edit rate <30%:",
    (eligible_users["edit_rate"] < 0.30).sum()
)
print(
    "Eligible users with edit rate <15%:",
    (eligible_users["edit_rate"] < 0.15).sum()
)

# 8. Save no files and change no original data; show sample flagged tasks
display(
    df.loc[
        df["final_flag_for_review"],
        [
            "user_id",
            "duration_clean",
            "task_time_clean",
            "ter",
            "analysis_char_count",
            "user_task_cps",
            "final_flag_shorter_than_audio",
            "final_flag_extreme_ter",
            "final_flag_extreme_cps"
        ]
    ].head(10)
)
```

```python

from difflib import SequenceMatcher

def text_similarity(user_text, whisper_text):
    if not isinstance(user_text, str) or not isinstance(whisper_text, str):
        return None
    if not whisper_text.strip():
        return None

    user_text = " ".join(user_text.split())
    whisper_text = " ".join(whisper_text.split())

    return SequenceMatcher(None, user_text, whisper_text).ratio()

fast_tasks["whisper_similarity"] = fast_tasks.apply(
    lambda row: text_similarity(row["user_text"], row["whisper_text"]),
    axis=1
)

fast_tasks["whisper_difference_rate"] = (
    1 - fast_tasks["whisper_similarity"]
)

display(
    fast_tasks[
        [
            "user_id", "ter", "whisper_similarity",
            "whisper_difference_rate", "user_text", "whisper_text"
        ]
    ].sort_values("whisper_difference_rate", ascending=False).head(20)
)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# ==================================================
# 1. LOAD DATA
# ==================================================
DATA_PATH = "../data/raw/transcription_data.csv"
OUTPUT_DIR = Path("../outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA_PATH)

required_cols = [
    "user_id",
    "whisper_text",
    "user_text",
    "is_edited",
    "duration",
    "time_taken_by_user",
]

missing_cols = [col for col in required_cols if col not in df.columns]
if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

# ==================================================
# 2. CLEAN DATA AND CALCULATE METRICS
# ==================================================
df["duration"] = pd.to_numeric(df["duration"], errors="coerce")
df["time_taken_by_user"] = pd.to_numeric(
    df["time_taken_by_user"], errors="coerce"
)

df["whisper_text"] = df["whisper_text"].fillna("").astype(str).str.strip()
df["user_text"] = df["user_text"].fillna("").astype(str).str.strip()

df["is_blank"] = (
    df["user_text"].str.lower().isin(["", "[blank]"])
)

# TER here means task time / audio duration.
df["ter"] = np.where(
    (df["duration"] > 0) & (df["time_taken_by_user"] >= 0),
    df["time_taken_by_user"] / df["duration"],
    np.nan,
)

# CPS = non-whitespace transcript characters / task time.
df["transcript_chars"] = df["user_text"].apply(
    lambda text: 0 if text.lower() == "[blank]"
    else sum(not char.isspace() for char in text)
)

df["cps"] = np.where(
    df["time_taken_by_user"] > 0,
    df["transcript_chars"] / df["time_taken_by_user"],
    np.nan,
)

# ==================================================
# 3. TASK-LEVEL WARNING FLAGS
# ==================================================
MIN_DURATION_TER = 1
MIN_DURATION_CPS = 5

EXTREME_TER_THRESHOLD = 0.1787571047
HIGH_CPS_THRESHOLD = 59.858

df["blank_flag"] = (
    df["is_blank"]
    & (df["whisper_text"] != "")
    & (df["duration"] > 1)
)

df["extreme_ter_flag"] = (
    (df["ter"] < EXTREME_TER_THRESHOLD)
    & (df["duration"] > MIN_DURATION_TER)
)

df["ter_below_1_flag"] = (
    (df["ter"] < 1)
    & (df["duration"] > MIN_DURATION_TER)
)

df["high_cps_flag"] = (
    (df["cps"] > HIGH_CPS_THRESHOLD)
    & (df["duration"] > MIN_DURATION_CPS)
)

# ==================================================
# 4. AGGREGATE BY TRANSCRIBER
# ==================================================
# Use eligible tasks as denominators for each warning rate.
df["blank_eligible"] = (
    (df["duration"] > 1)
    & (df["whisper_text"] != "")
)

df["ter_eligible"] = (
    (df["duration"] > MIN_DURATION_TER)
    & df["ter"].notna()
)

df["cps_eligible"] = (
    (df["duration"] > MIN_DURATION_CPS)
    & df["cps"].notna()
)

users = df.groupby("user_id", dropna=False).agg(
    total_tasks=("user_id", "size"),

    blank_tasks=("blank_flag", "sum"),
    blank_eligible_tasks=("blank_eligible", "sum"),

    extreme_ter_tasks=("extreme_ter_flag", "sum"),
    ter_below_1_tasks=("ter_below_1_flag", "sum"),
    ter_eligible_tasks=("ter_eligible", "sum"),

    high_cps_tasks=("high_cps_flag", "sum"),
    cps_eligible_tasks=("cps_eligible", "sum"),
).reset_index()

# Rates are percentages from 0 to 100.
def percentage(numerator, denominator):
    return np.where(
        denominator > 0,
        numerator / denominator * 100,
        0.0,
    )

users["blank_rate"] = percentage(
    users["blank_tasks"], users["blank_eligible_tasks"]
)

users["extreme_ter_rate"] = percentage(
    users["extreme_ter_tasks"], users["ter_eligible_tasks"]
)

users["ter_below_1_rate"] = percentage(
    users["ter_below_1_tasks"], users["ter_eligible_tasks"]
)

users["high_cps_rate"] = percentage(
    users["high_cps_tasks"], users["cps_eligible_tasks"]
)

# ==================================================
# 5. BLOCK-CANDIDATE RULE
# ==================================================
MIN_TASKS = 20
BLANK_RATE_THRESHOLD = 2.0
TER_BELOW_1_RATE_THRESHOLD = 30.0

users["block_candidate"] = (
    (users["total_tasks"] >= MIN_TASKS)
    & (users["blank_rate"] > BLANK_RATE_THRESHOLD)
    & (users["ter_below_1_rate"] > TER_BELOW_1_RATE_THRESHOLD)
)

# Explain why each user is flagged.
users["candidate_reason"] = np.select(
    [
        users["block_candidate"],
        users["total_tasks"] < MIN_TASKS,
    ],
    [
        "High blank rate + high TER<1 rate",
        "Insufficient task history",
    ],
    default="Does not meet blocking rule",
)

# ==================================================
# 6. DISPLAY RESULTS
# ==================================================
display_cols = [
    "user_id",
    "total_tasks",
    "blank_tasks",
    "blank_eligible_tasks",
    "blank_rate",
    "ter_below_1_tasks",
    "ter_eligible_tasks",
    "ter_below_1_rate",
    "extreme_ter_tasks",
    "extreme_ter_rate",
    "high_cps_tasks",
    "high_cps_rate",
    "block_candidate",
    "candidate_reason",
]

users = users.sort_values(
    ["block_candidate", "total_tasks"],
    ascending=[False, False],
)

print("Total tasks:", len(df))
print("Unique transcribers:", df["user_id"].nunique())
print("Block candidates:", int(users["block_candidate"].sum()))

display(users[display_cols].head(50))

# ==================================================
# 7. SAVE OUTPUTS
# ==================================================
users[display_cols].to_csv(
    OUTPUT_DIR / "transcriber_risk_summary.csv",
    index=False,
)

users.loc[users["block_candidate"], display_cols].to_csv(
    OUTPUT_DIR / "block_candidates.csv",
    index=False,
)

df.to_csv(
    OUTPUT_DIR / "task_level_warning_flags.csv",
    index=False,
)

print("\nSaved files:")
print(OUTPUT_DIR / "transcriber_risk_summary.csv")
print(OUTPUT_DIR / "block_candidates.csv")
print(OUTPUT_DIR / "task_level_warning_flags.csv")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# =========================================================
# 1. CONFIGURATION
# =========================================================
DATA_PATH = "../data/raw/transcription_data.csv"
OUTPUT_DIR = Path("../outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

MIN_TASKS = 5

# Existing task-level thresholds
EXTREME_TER_THRESHOLD = 0.1787571047
HIGH_CPS_THRESHOLD = 59.858

# Transcriber-level thresholds (percentages)
BLANK_RATE_THRESHOLD = 2.0
TER_BELOW_1_RATE_THRESHOLD = 30.0

# =========================================================
# 2. LOAD DATA
# =========================================================
df = pd.read_csv(DATA_PATH)

required_cols = [
    "user_id",
    "whisper_text",
    "user_text",
    "is_edited",
    "duration",
    "time_taken_by_user",
]

missing_cols = [c for c in required_cols if c not in df.columns]
if missing_cols:
    raise ValueError(f"Missing required columns: {missing_cols}")

df["duration"] = pd.to_numeric(df["duration"], errors="coerce")
df["time_taken_by_user"] = pd.to_numeric(
    df["time_taken_by_user"], errors="coerce"
)

for col in ["whisper_text", "user_text"]:
    df[col] = df[col].fillna("").astype(str).str.strip()

# =========================================================
# 3. TASK-LEVEL METRICS
# =========================================================
df["is_blank"] = df["user_text"].str.lower().isin(["", "[blank]"])

# TER used here = task time / audio duration
df["ter"] = np.where(
    (df["duration"] > 0) & (df["time_taken_by_user"] >= 0),
    df["time_taken_by_user"] / df["duration"],
    np.nan,
)

# CPS = non-whitespace transcript characters / task time
df["transcript_chars"] = df["user_text"].apply(
    lambda text: (
        0 if text.lower() == "[blank]"
        else sum(not char.isspace() for char in text)
    )
)

df["cps"] = np.where(
    df["time_taken_by_user"] > 0,
    df["transcript_chars"] / df["time_taken_by_user"],
    np.nan,
)

# =========================================================
# 4. DEFINE WARNING SIGNALS
# =========================================================
df["blank_eligible"] = (
    (df["duration"] > 1)
    & (df["whisper_text"] != "")
)

df["ter_eligible"] = (
    (df["duration"] > 1)
    & df["ter"].notna()
)

df["cps_eligible"] = (
    (df["duration"] > 5)
    & df["cps"].notna()
)

df["blank_flag"] = (
    df["is_blank"]
    & (df["whisper_text"] != "")
    & (df["duration"] > 1)
)

df["extreme_ter_flag"] = (
    (df["ter"] < EXTREME_TER_THRESHOLD)
    & (df["duration"] > 1)
)

df["ter_below_1_flag"] = (
    (df["ter"] < 1)
    & (df["duration"] > 1)
)

df["high_cps_flag"] = (
    (df["cps"] > HIGH_CPS_THRESHOLD)
    & (df["duration"] > 5)
)

# =========================================================
# 5. AGGREGATE BY TRANSCRIBER
# =========================================================
users = df.groupby("user_id", dropna=False).agg(
    total_tasks=("user_id", "size"),

    blank_tasks=("blank_flag", "sum"),
    blank_eligible_tasks=("blank_eligible", "sum"),

    extreme_ter_tasks=("extreme_ter_flag", "sum"),
    ter_below_1_tasks=("ter_below_1_flag", "sum"),
    ter_eligible_tasks=("ter_eligible", "sum"),

    high_cps_tasks=("high_cps_flag", "sum"),
    cps_eligible_tasks=("cps_eligible", "sum"),
).reset_index()


def percentage(count, eligible):
    return np.where(
        eligible > 0,
        count / np.maximum(eligible, 1) * 100,
        0.0,
    )


users["blank_rate"] = percentage(
    users["blank_tasks"],
    users["blank_eligible_tasks"],
)

users["extreme_ter_rate"] = percentage(
    users["extreme_ter_tasks"],
    users["ter_eligible_tasks"],
)

users["ter_below_1_rate"] = percentage(
    users["ter_below_1_tasks"],
    users["ter_eligible_tasks"],
)

users["high_cps_rate"] = percentage(
    users["high_cps_tasks"],
    users["cps_eligible_tasks"],
)

# =========================================================
# 6. RELAXED BLOCK-CANDIDATE RULE
# =========================================================
# Four warning signals:
# 1. Blank rate > 2%
# 2. Extreme TER rate > 0%
# 3. High CPS rate > 0%
# 4. TER-below-1 rate > 30%
#
# Candidate requires at least 5 tasks and 2+ signals.
# These are review candidates, not confirmed fraud.

users["blank_signal"] = (
    users["blank_rate"] > BLANK_RATE_THRESHOLD
)

users["extreme_ter_signal"] = (
    users["extreme_ter_rate"] > 0
)

users["high_cps_signal"] = (
    users["high_cps_rate"] > 0
)

users["ter_below_1_signal"] = (
    users["ter_below_1_rate"] > TER_BELOW_1_RATE_THRESHOLD
)

signal_cols = [
    "blank_signal",
    "extreme_ter_signal",
    "high_cps_signal",
    "ter_below_1_signal",
]

users["warning_signal_count"] = (
    users[signal_cols].sum(axis=1)
)

users["block_candidate"] = (
    (users["total_tasks"] >= MIN_TASKS)
    & (users["warning_signal_count"] >= 2)
)

users["candidate_reason"] = users.apply(
    lambda row: ", ".join(
        label
        for col, label in [
            ("blank_signal", "High blank rate"),
            ("extreme_ter_signal", "Extreme TER"),
            ("high_cps_signal", "High CPS"),
            ("ter_below_1_signal", "High TER<1 rate"),
        ]
        if row[col]
    ) if row["block_candidate"] else "Does not meet rule",
    axis=1,
)

# =========================================================
# 7. SUMMARY AND RESULTS
# =========================================================
result_cols = [
    "user_id",
    "total_tasks",
    "blank_tasks",
    "blank_eligible_tasks",
    "blank_rate",
    "ter_below_1_tasks",
    "ter_eligible_tasks",
    "ter_below_1_rate",
    "extreme_ter_tasks",
    "extreme_ter_rate",
    "high_cps_tasks",
    "cps_eligible_tasks",
    "high_cps_rate",
    "warning_signal_count",
    "candidate_reason",
    "block_candidate",
]

users = users.sort_values(
    ["block_candidate", "warning_signal_count", "total_tasks"],
    ascending=[False, False, False],
)

print("=" * 55)
print("TRANSCRIBER RISK SUMMARY")
print("=" * 55)
print(f"Total task rows: {len(df):,}")
print(f"Unique transcribers: {df['user_id'].nunique():,}")
print(f"Transcribers with >= {MIN_TASKS} tasks: "
      f"{(users['total_tasks'] >= MIN_TASKS).sum():,}")
print(f"Block candidates for review: "
      f"{users['block_candidate'].sum():,}")

print("\nWarning-signal counts:")
for col in signal_cols:
    print(f"{col}: {int(users[col].sum()):,}")

print("\nCandidate list:")
display(users.loc[users["block_candidate"], result_cols])

print("\nTop 30 transcribers by warning signals:")
display(users[result_cols].head(30))

# =========================================================
# 8. SAVE OUTPUT FILES
# =========================================================
users[result_cols].to_csv(
    OUTPUT_DIR / "transcriber_risk_summary.csv",
    index=False,
)

users.loc[users["block_candidate"], result_cols].to_csv(
    OUTPUT_DIR / "block_candidates.csv",
    index=False,
)

df.to_csv(
    OUTPUT_DIR / "task_level_warning_flags.csv",
    index=False,
)

print("\nFiles saved:")
print(OUTPUT_DIR / "transcriber_risk_summary.csv")
print(OUTPUT_DIR / "block_candidates.csv")
print(OUTPUT_DIR / "task_level_warning_flags.csv")
```
